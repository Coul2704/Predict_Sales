```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Chargement des données 
df =pd.read_csv("Auto Sales data.csv")
```

```python
#CA par gamme produit
productline_sales = df.groupby("PRODUCTLINE")["SALES"].sum().sort_values(ascending=False)

productline_sales.plot(kind="bar", figsize=(10,5))
plt.title("Chiffre d'affaires par gamme produit")
plt.xticks(rotation=45)
plt.show()
```

```python
#Quantité vendue par gamme
productline_qty = df.groupby("PRODUCTLINE")["QUANTITYORDERED"].sum().sort_values(ascending=False)

productline_qty.plot(kind="bar", figsize=(10,5))
plt.title("Quantité vendue par gamme produit")
plt.xticks(rotation=45)
plt.show()
```

```python
#Top 10 produits par chiffre d’affaires
top_products_sales = df.groupby("PRODUCTLINE")["SALES"].sum().sort_values(ascending=False).head(10)

top_products_sales.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 produits par chiffre d'affaires")
plt.xticks(rotation=45)
plt.show()
```

```python
#Produits les plus vendus
top_products_qty = df.groupby("PRODUCTLINE")["QUANTITYORDERED"].sum().sort_values(ascending=False).head(10)

top_products_qty.plot(kind="bar", figsize=(10,5))
plt.title("Top 10 produits par quantité vendue")
plt.xticks(rotation=45)
plt.show()
```

```python
#Prix moyen par gamme produit
avg_price_productline = df.groupby("PRODUCTLINE")["PRICEEACH"].mean().sort_values(ascending=False)

avg_price_productline.plot(kind="bar", figsize=(10,5))
plt.title("Prix moyen par gamme produit")
plt.xticks(rotation=45)
plt.show()
```

```python
#Relation prix vs ventes
plt.scatter(df["PRICEEACH"], df["SALES"])
plt.title("Relation entre prix unitaire et ventes")
plt.xlabel("Prix unitaire")
plt.ylabel("Ventes")
plt.show()
```

```python
#Contribution des produits au CA total
product_share = df.groupby("PRODUCTLINE")["SALES"].sum()
product_share = product_share / product_share.sum() * 100

product_share.plot(kind="pie", autopct="%1.1f%%", figsize=(7,7))
plt.title("Part de contribution des gammes au CA")
plt.ylabel("")
plt.show()
```

```python
#Analyse des performances produit
product_summary = df.groupby("PRODUCTLINE").agg({
    "SALES": "sum",
    "QUANTITYORDERED": "sum",
    "PRICEEACH": "mean"
}).sort_values("SALES", ascending=False)

product_summary
```
